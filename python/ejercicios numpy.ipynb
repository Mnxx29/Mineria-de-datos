Ejercicio N9

```python
#El Producto Matricial: Crea una matriz A de 2 X 3 y una matriz B de 3 X 2. Realiza el producto matricial (@) y explica por qué el 
#resultado es una matriz de 2 X 2. 

A=np.array([[1,2,3],[4,5,6]])
B=np.array([[1,2],[3,4],[5,6]])

print(A@B)
```

Ejericicio N10


```python
#10. La Transpuesta: Crea una matriz de $4 \times 2$ con valores del 1 al 8. Aplica la función .T (transpuesta) y
#observa cómo cambian las dimensiones de la matriz.

A=np.array([[1,2],[3,4],[5,6],[7,8]])
print("Matriz A= \n",A)
print("Matriz A transpuesta= \n",A.T)
print("Dimensiones de la matriz A= ",A.shape)
print("Dimensiones de la matriz A transpuesta= ",A.T.shape)

```
